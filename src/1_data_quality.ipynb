```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt


#Load the dataset
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset['train'].to_pandas()

#Data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.head(10)
```

```python
df.info()
```

```python
# Figuring out the percentage of missing values in each column

missing = (df.isna().mean().sort_values(ascending=False).mul(100).round(2))

missing
```

```python
df["has_salary"] = df["salary_year_avg"].notna()
```

```python
# What roles disclose salary most often?

df.groupby('job_title_short')['has_salary'].mean().sort_values(ascending=False)
```

```python
# Which countries report salary most often?

df.groupby('job_country')['has_salary'].mean().sort_values(ascending=False).head(10)
```

```python
# Identifying duplicates

df.duplicated(subset=['company_name', 'job_title', 'job_location']).mean()
```

```python
df.groupby(["company_name", "job_title", "job_location"]).size().sort_values(ascending=False)
```
